````python
import pandas as pd
import numpy as np
from pathlib import Path

# ── Chemins ──────────────────────────────────────────────────────────────────
RAW_DIR    = Path("../data/01_raw")
OUTPUT_DIR = Path("../data/02_interim")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ── Mapping colonnes par fichier ──────────────────────────────────────────────
FILE_COLS = {
    "chest_physiology_summary.csv": {
        "hr":      "chest_physiology_summary_hr",
        "br":      "chest_physiology_summary_br",
        "posture": "chest_physiology_summary_posture",
        "hrv":     "chest_physiology_summary_hrv",},
    "wrist_acc.csv":              {"ax": "wrist_acc_ax", "ay": "wrist_acc_ay", "az": "wrist_acc_az"},
    "wrist_eda.csv":              {"eda": "wrist_eda_eda"},
    "wrist_hr.csv":               {"hr": "wrist_hr_hr"},
    "wrist_ibi.csv":              {"duration": "wrist_ibi_duration"},
    "wrist_skin_temperature.csv": {"temp": "wrist_skin_temperature_temp"},
}

# ── Labels d'intérêt dans exp_markers.csv ────────────────────────────────────
LABELS_OF_INTEREST = ["baseline", "activity", "fatigue"]

# ── Fonction : extraire les intervalles de labels depuis exp_markers.csv ──────
def get_label_intervals(markers_path):
    df = pd.read_csv(markers_path)
    df.columns = df.columns.str.strip()

    # Colonnes réelles
    ts_col    = "utcTime"
    event_col = "eventMarker"

    intervals = []
    active_labels = {}  # label -> start_time

    for _, row in df.iterrows():
        ts    = float(row[ts_col])
        event = str(row[event_col]).strip().lower()

        for lbl in LABELS_OF_INTEREST:
            if f"start_{lbl}" in event:
                active_labels[lbl] = ts
            elif f"end_{lbl}" in event:
                if lbl in active_labels:
                    intervals.append((active_labels.pop(lbl), ts, lbl))

    print(f"  ℹ Intervalles trouvés: {intervals}")
    return intervals

# ── Fonction : charger un CSV et renommer les colonnes ────────────────────────
def load_and_rename(filepath, col_map):
    try:
        df = pd.read_csv(filepath)
        df.columns = df.columns.str.strip()

        # Détecter automatiquement la colonne timestamp
        ts_candidates = [c for c in df.columns if "time" in c.lower() or "timestamp" in c.lower()]
        if not ts_candidates:
            print(f"  ⚠ Pas de colonne timestamp dans {filepath.name} | colonnes: {df.columns.tolist()}")
            return None
        ts_col = ts_candidates[0]
        df = df.rename(columns={ts_col: "timestamp"})

        # Colonnes en minuscules pour le mapping
        df.columns = df.columns.str.lower()

        cols_to_keep = {"timestamp": "timestamp"}
        for src, dst in col_map.items():
            src_l = src.lower()
            if src_l in df.columns:
                cols_to_keep[src_l] = dst

        df = df[list(cols_to_keep.keys())].rename(columns=cols_to_keep)
        df["timestamp"] = pd.to_numeric(df["timestamp"], errors="coerce")
        return df.dropna(subset=["timestamp"])

    except Exception as e:
        print(f"  ⚠ Erreur lecture {filepath}: {e}")
        return None

# ── Fonction : assigner un label à chaque timestamp ───────────────────────────
def assign_labels(df, intervals):
    df = df.copy()
    df["label"] = None  # utiliser None au lieu de np.nan pour accepter les strings
    df["label"] = df["label"].astype(object)

    for start, stop, lbl in intervals:
        mask = (df["timestamp"] >= start) & (df["timestamp"] <= stop)
        df.loc[mask, "label"] = lbl

    return df.dropna(subset=["label"])  # supprimer les lignes hors intervalles (les labels manquante a supprimé)

# ── Boucle principale ─────────────────────────────────────────────────────────
all_data = []

for participant_dir in sorted(RAW_DIR.iterdir()):
    if not participant_dir.is_dir():
        continue
    participant_id = participant_dir.name  # ex: "01", "02", ...

    for session_dir in sorted(participant_dir.iterdir()):
        if not session_dir.is_dir():
            continue
        session_id = session_dir.name  # ex: "01", "02", "03"

        print(f"▶ Participant {participant_id} | Session {session_id}")
        #-------------------------------------------------------------------------------- >>      
        # 1. Charger les labels
        markers_path = session_dir / "exp_markers.csv"
        if not markers_path.exists():
            print(f"  ⚠ exp_markers.csv manquant, session ignorée")
            continue

        intervals = get_label_intervals(markers_path)
        if not intervals:
            print(f"  ⚠ Aucun intervalle trouvé dans exp_markers.csv")
            continue

        # 2. Charger et merger tous les fichiers de signaux
        merged = None
        for filename, col_map in FILE_COLS.items():
            filepath = session_dir / filename
            if not filepath.exists():
                print(f"  ⚠ {filename} manquant")
                continue

            df_sig = load_and_rename(filepath, col_map)
            if df_sig is None or df_sig.empty:
                continue

            if merged is None:
                merged = df_sig
            else:
                merged = pd.merge_asof(
                    merged.sort_values("timestamp"),
                    df_sig.sort_values("timestamp"),
                    on="timestamp",
                    direction="nearest",
                    tolerance=1000  # 1 seconde de tolérance
                )

        if merged is None or merged.empty:
            print(f"  ⚠ Aucune donnée mergée")
            continue

        # 3. Assigner les labels et filtrer
        merged = assign_labels(merged, intervals)
        if merged.empty:
            print(f"  ⚠ Aucune donnée après filtrage des labels")
            continue

        # 4. Ajouter participant_id et session_id
        merged.insert(1, "participant_id", participant_id)
        merged.insert(2, "session_id", session_id)

        print(f"  ✔ {len(merged)} lignes | labels: {merged['label'].value_counts().to_dict()}")
        all_data.append(merged)

# ── Concaténer et sauvegarder ─────────────────────────────────────────────────
if all_data:
    final_df = pd.concat(all_data, ignore_index=True)

    # S'assurer de l'ordre des colonnes
    expected_cols = [
        "timestamp", "participant_id", "session_id",
        "chest_physiology_summary_hr", "chest_physiology_summary_br",
        "chest_physiology_summary_posture", "chest_physiology_summary_hrv",
        "wrist_acc_ax", "wrist_acc_ay", "wrist_acc_az",
        "wrist_eda_eda", "wrist_hr_hr", "wrist_ibi_duration",
        "wrist_skin_temperature_temp", "label"
    ]
    for col in expected_cols:
        if col not in final_df.columns:
            final_df[col] = np.nan

    final_df = final_df[expected_cols]

    output_path = OUTPUT_DIR / "brut_dataset1.csv"
    final_df.to_csv(output_path, index=False)

    print(f"\n✅ Dataset sauvegardé : {output_path}")
    print(f"   Lignes totales     : {len(final_df)}")
    print(f"   Participants       : {final_df['participant_id'].nunique()}")
    print(f"   Distribution labels:\n{final_df['label'].value_counts()}")
else:
    print("❌ Aucune donnée collectée.")
````

````python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.signal import butter, filtfilt, find_peaks

# ── Chemins ───────────────────────────────────────────────────────────────────
RAW_DIR    = Path("../data/01_raw")
OUTPUT_DIR = Path("../data/02_interim")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ── Paramètres respiratoires ──────────────────────────────────────────────────
FS_BREATHING = 25.0  # Hz
FMIN         = 0.05  # Hz
FMAX         = 0.80  # Hz
MIN_DIST_S   = 1.5   # secondes minimum entre deux pics
PROM_MIN     = 100   # prominence minimale
PROM_OK      = 500   # prominence = bon pic

# ── Fréquences natives des capteurs ──────────────────────────────────────────
SENSOR_FREQ = {
    "chest_physiology_summary.csv": 1,
    "wrist_acc.csv":                32,
    "wrist_eda.csv":                4,
    "wrist_hr.csv":                 1,
    "wrist_ibi.csv":                0.59,
    "wrist_skin_temperature.csv":   4,
}

TARGET_FREQ   = 4     # Hz cible
TARGET_PERIOD = 250   # ms (= 1000 / 4Hz)

# ── Mapping colonnes ──────────────────────────────────────────────────────────
FILE_COLS = {
    "chest_physiology_summary.csv": {
        "hr":      "chest_physiology_summary_hr",
        "br":      "chest_physiology_summary_br",
        "posture": "chest_physiology_summary_posture",
        "hrv":     "chest_physiology_summary_hrv",
    },
    "wrist_acc.csv":              {"ax": "wrist_acc_ax", "ay": "wrist_acc_ay", "az": "wrist_acc_az"},
    "wrist_eda.csv":              {"eda": "wrist_eda_eda"},
    "wrist_hr.csv":               {"hr": "wrist_hr_hr"},
    "wrist_ibi.csv":              {"duration": "wrist_ibi_duration"},
    "wrist_skin_temperature.csv": {"temp": "wrist_skin_temperature_temp"},
}

LABELS_OF_INTEREST = ["baseline", "activity", "fatigue"]

# ═══════════════════════════════════════════════════════════════════════════════
# FONCTION 1 : Extraire les intervalles de labels
# ═══════════════════════════════════════════════════════════════════════════════
def get_label_intervals(markers_path):
    df = pd.read_csv(markers_path)
    df.columns = df.columns.str.strip()

    intervals    = []
    active_labels = {}

    for _, row in df.iterrows():
        ts    = float(row["utcTime"])
        event = str(row["eventMarker"]).strip().lower()

        for lbl in LABELS_OF_INTEREST:
            if f"start_{lbl}" in event:
                active_labels[lbl] = ts
            elif f"end_{lbl}" in event:
                if lbl in active_labels:
                    intervals.append((active_labels.pop(lbl), ts, lbl))

    print(f"  ℹ Intervalles trouvés: {intervals}")
    return intervals

# ═══════════════════════════════════════════════════════════════════════════════
# FONCTION 2 : Calcul RPM depuis chest_raw_breathing.csv
# ═══════════════════════════════════════════════════════════════════════════════
def compute_breathing_rpm(filepath):
    """
    Lit chest_raw_breathing.csv, applique un filtre bandpass,
    détecte les pics et calcule le RPM pic à pic.
    Retourne un DataFrame à fréquence irrégulière :
        timestamp (ms) | chest_raw_breathing_rpm
    """
    try:
        df  = pd.read_csv(filepath)
        df.columns = df.columns.str.strip()

        ts  = df["timestamp"].values.astype(np.int64)
        sig = df["breathing_waveform"].values.astype(float)

        # ── Étape 1 : Filtre bandpass Butterworth ────────────────────────────
        sig_c    = sig - np.mean(sig)                          # centrer
        b, a     = butter(2, [FMIN, FMAX], btype='band', fs=FS_BREATHING)
        sig_filt = filtfilt(b, a, sig_c)                       # filtrer

        # ── Étape 2 : Détecter les pics ──────────────────────────────────────
        min_dist_samples = int(MIN_DIST_S * FS_BREATHING)      # 1.5s × 25Hz = 37
        peaks, props = find_peaks(
            sig_filt,
            distance=min_dist_samples,
            prominence=PROM_MIN
        )

        if len(peaks) < 2:
            print(f"  ⚠ Pas assez de pics détectés dans {filepath.name}")
            return None

        # ── Étape 3 : Calculer RPM pic à pic ─────────────────────────────────
        rows = []
        for i in range(len(peaks)):
            idx = peaks[i]

            if i < len(peaks) - 1:
                next_idx = peaks[i + 1]
                period_s = (next_idx - idx) / FS_BREATHING
                br_rpm   = round(60.0 / period_s, 2)
            else:
                br_rpm   = np.nan  # dernier pic → pas de suivant

            prom    = float(props["prominences"][i])
            quality = 1 if prom >= PROM_OK else 0

            rows.append({
                "timestamp":                  int(ts[idx]),
                "chest_raw_breathing_rpm":    br_rpm,
                "chest_raw_breathing_quality": quality
            })

        df_rpm = pd.DataFrame(rows).dropna(subset=["chest_raw_breathing_rpm"])

        # Valider les RPM (filtrer les valeurs aberrantes)
        df_rpm = df_rpm[
            (df_rpm["chest_raw_breathing_rpm"] >= 3) &
            (df_rpm["chest_raw_breathing_rpm"] <= 60)
        ]

        print(f"  ℹ Breathing RPM : "
              f"min={df_rpm['chest_raw_breathing_rpm'].min():.1f} "
              f"max={df_rpm['chest_raw_breathing_rpm'].max():.1f} "
              f"moy={df_rpm['chest_raw_breathing_rpm'].mean():.1f} "
              f"pics={len(df_rpm)}")

        return df_rpm

    except Exception as e:
        print(f"  ⚠ Erreur breathing RPM {filepath.name}: {e}")
        return None

# ═══════════════════════════════════════════════════════════════════════════════
# FONCTION 3 : Resampler un DataFrame signal vers 4Hz
# ═══════════════════════════════════════════════════════════════════════════════
def resample_to_4hz(df, native_freq, target_period_ms=250):
    """
    Resamples un DataFrame (index = timestamp ms) vers 4Hz.
    - native_freq > 4  : downsample par moyenne
    - native_freq < 1  : irrégulier (IBI, RPM) → forward fill
    - native_freq < 4  : upsample par interpolation linéaire
    - native_freq = 4  : réindexer proprement
    """
    # Convertir timestamp ms → DatetimeIndex
    df["datetime"] = pd.to_datetime(df["timestamp"], unit="ms")
    df = df.set_index("datetime").drop(columns=["timestamp"])

    period_str = f"{target_period_ms}ms"  # "250ms"

    if native_freq > TARGET_FREQ:
        # 32Hz → 4Hz : moyenne sur 8 points
        df_out = df.resample(period_str).mean()

    elif native_freq < 1:
        # IBI (0.59Hz) ou RPM irrégulier → forward fill
        df_out = df.resample(period_str).ffill()

    elif native_freq < TARGET_FREQ:
        # 1Hz → 4Hz : interpolation linéaire
        df_out = df.resample(period_str).interpolate(method="linear")

    else:
        # 4Hz → 4Hz : réindexer proprement
        df_out = df.resample(period_str).mean()

    # Remettre timestamp ms comme colonne
    df_out["timestamp"] = (
        df_out.index.astype(np.int64) // 10**6
    )
    return df_out.reset_index(drop=True)

# ═══════════════════════════════════════════════════════════════════════════════
# FONCTION 4 : Charger, renommer et resampler un fichier signal
# ═══════════════════════════════════════════════════════════════════════════════
def load_rename_resample(filepath, col_map, native_freq):
    try:
        df = pd.read_csv(filepath)
        df.columns = df.columns.str.strip()

        # Détecter la colonne timestamp
        ts_candidates = [c for c in df.columns if "time" in c.lower()]
        if not ts_candidates:
            print(f"  ⚠ Pas de colonne timestamp dans {filepath.name}")
            return None

        df = df.rename(columns={ts_candidates[0]: "timestamp"})
        df.columns = df.columns.str.lower()

        # Garder uniquement les colonnes utiles
        cols_to_keep = {"timestamp": "timestamp"}
        for src, dst in col_map.items():
            if src.lower() in df.columns:
                cols_to_keep[src.lower()] = dst

        df = df[list(cols_to_keep.keys())].rename(columns=cols_to_keep)
        df["timestamp"] = pd.to_numeric(df["timestamp"], errors="coerce")
        df = df.dropna(subset=["timestamp"]).sort_values("timestamp")

        # Resampler vers 4Hz
        df = resample_to_4hz(df, native_freq)

        return df

    except Exception as e:
        print(f"  ⚠ Erreur {filepath.name}: {e}")
        return None

# ═══════════════════════════════════════════════════════════════════════════════
# FONCTION 5 : Assigner les labels
# ═══════════════════════════════════════════════════════════════════════════════
def assign_labels(df, intervals):
    df = df.copy()
    df["label"] = None
    df["label"] = df["label"].astype(object)

    for start, stop, lbl in intervals:
        mask = (df["timestamp"] >= start) & (df["timestamp"] <= stop)
        df.loc[mask, "label"] = lbl

    return df.dropna(subset=["label"])

# ═══════════════════════════════════════════════════════════════════════════════
# BOUCLE PRINCIPALE
# ═══════════════════════════════════════════════════════════════════════════════
all_data = []

for participant_dir in sorted(RAW_DIR.iterdir()):
    if not participant_dir.is_dir():
        continue
    participant_id = participant_dir.name

    for session_dir in sorted(participant_dir.iterdir()):
        if not session_dir.is_dir():
            continue
        session_id = session_dir.name

        print(f"\n▶ Participant {participant_id} | Session {session_id}")

        # ── 1. Labels ────────────────────────────────────────────────────────
        markers_path = session_dir / "exp_markers.csv"
        if not markers_path.exists():
            print(f"  ⚠ exp_markers.csv manquant")
            continue

        intervals = get_label_intervals(markers_path)
        if not intervals:
            print(f"  ⚠ Aucun intervalle trouvé")
            continue
        print("Signal min/max:", merged["timestamp"].min(), merged["timestamp"].max())
        print("Labels min/max:", intervals[0][0], intervals[-1][1])

        # ── 2. Signaux standards ─────────────────────────────────────────────
        merged = None
        for filename, col_map in FILE_COLS.items():
            filepath = session_dir / filename
            if not filepath.exists():
                print(f"  ⚠ {filename} manquant")
                continue

            native_freq = SENSOR_FREQ[filename]
            df_sig = load_rename_resample(filepath, col_map, native_freq)

            if df_sig is None or df_sig.empty:
                continue

            if merged is None:
                merged = df_sig
            else:
                merged = pd.merge_asof(
                    merged.sort_values("timestamp"),
                    df_sig.sort_values("timestamp"),
                    on="timestamp",
                    direction="nearest",
                    tolerance=TARGET_PERIOD  # 250ms
                )

        # ── 3. Breathing RPM ─────────────────────────────────────────────────
        breathing_path = session_dir / "chest_raw_breathing.csv"
        if breathing_path.exists():
            df_rpm = compute_breathing_rpm(breathing_path)

            if df_rpm is not None and not df_rpm.empty:
                # Resampler le RPM irrégulier → 4Hz (forward fill)
                df_rpm = resample_to_4hz(df_rpm, native_freq=0.59)

                if merged is None:
                    merged = df_rpm
                else:
                    merged = pd.merge_asof(
                        merged.sort_values("timestamp"),
                        df_rpm.sort_values("timestamp"),
                        on="timestamp",
                        direction="nearest",
                        tolerance=TARGET_PERIOD
                    )
        else:
            print(f"  ⚠ chest_raw_breathing.csv manquant")

        if merged is None or merged.empty:
            print(f"  ⚠ Aucune donnée mergée")
            continue

        # ── 4. Labels ────────────────────────────────────────────────────────
        merged = assign_labels(merged, intervals)
        if merged.empty:
            print(f"  ⚠ Aucune donnée après filtrage labels")
            continue

        # ── 5. Ajouter participant_id et session_id ───────────────────────────
        merged.insert(1, "participant_id", participant_id)
        merged.insert(2, "session_id", session_id)

        print(f"  ✔ {len(merged)} lignes | "
              f"labels: {merged['label'].value_counts().to_dict()}")
        all_data.append(merged)

# ═══════════════════════════════════════════════════════════════════════════════
# SAUVEGARDE
# ═══════════════════════════════════════════════════════════════════════════════
if all_data:
    final_df = pd.concat(all_data, ignore_index=True)

    expected_cols = [
        "timestamp", "participant_id", "session_id",
        "chest_physiology_summary_hr",
        "chest_physiology_summary_br",
        "chest_physiology_summary_posture",
        "chest_physiology_summary_hrv",
        "wrist_acc_ax", "wrist_acc_ay", "wrist_acc_az",
        "wrist_eda_eda",
        "wrist_hr_hr",
        "wrist_ibi_duration",
        "wrist_skin_temperature_temp",
        "chest_raw_breathing_rpm",
        "chest_raw_breathing_quality",
        "label"
    ]

    for col in expected_cols:
        if col not in final_df.columns:
            final_df[col] = np.nan

    final_df = final_df[expected_cols]

    output_path = OUTPUT_DIR / "brut_dataset1.csv"
    final_df.to_csv(output_path, index=False)

    print(f"\n✅ Dataset sauvegardé : {output_path}")
    print(f"   Lignes totales      : {len(final_df)}")
    print(f"   Participants        : {final_df['participant_id'].nunique()}")
    print(f"   Fréquence           : 4Hz")
    print(f"   Colonnes            : {len(expected_cols)}")
    print(f"   Distribution labels :\n{final_df['label'].value_counts()}")
    print(f"\n   Aperçu NaN par colonne :")
    print(final_df[expected_cols].isna().sum().to_string())
else:
    print("❌ Aucune donnée collectée.")
````

````python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.signal import butter, filtfilt, find_peaks

# ── Chemins ───────────────────────────────────────────────────────────────────
RAW_DIR    = Path("../data/01_raw")
OUTPUT_DIR = Path("../data/02_interim")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
PATH_BREATHING = "chest_raw_breathing.csv"

# ── Paramètres respiration ────────────────────────────────────────────────────
FS_BREATHING = 25.0
FMIN, FMAX   = 0.05, 0.80
MIN_DIST_S   = 1.5
PROM_MIN     = 100
PROM_OK      = 500

# ── Fréquences capteurs ───────────────────────────────────────────────────────
SENSOR_FREQ = {
    "chest_physiology_summary.csv": 1,
    "wrist_acc.csv":                32,
    "wrist_eda.csv":                4,
    "wrist_hr.csv":                 1,
    "wrist_ibi.csv":                0.59,
    "wrist_skin_temperature.csv":   4,
}

TARGET_FREQ   = 4
TARGET_PERIOD = 250  # ms

# ── Colonnes ──────────────────────────────────────────────────────────────────
#    "chest_physiology_summary.csv": {"hr": "chest_hr","br": "chest_br","posture": "chest_posture","hrv": "chest_hrv",},
FILE_COLS = {

    "wrist_acc.csv": {"ax": "acc_x", "ay": "acc_y", "az": "acc_z"},
    "wrist_eda.csv": {"eda": "eda"},
    "wrist_hr.csv": {"hr": "wrist_hr"},
    "wrist_ibi.csv": {"duration": "ibi"},
    "wrist_skin_temperature.csv": {"temp": "temp"},
}

LABELS_OF_INTEREST = ["baseline", "activity", "fatigue"]

# ═══════════════════════════════════════════════════════════════════════════════
# 1. Labels → convertir en temps relatif
# ═══════════════════════════════════════════════════════════════════════════════
def get_label_intervals(markers_path):
    df = pd.read_csv(markers_path)
    df.columns = df.columns.str.strip()

    intervals = []
    active = {}

    for _, row in df.iterrows():
        ts = float(row["utcTime"])
        event = str(row["eventMarker"]).lower()

        for lbl in LABELS_OF_INTEREST:
            if f"start_{lbl}" in event:
                active[lbl] = ts
            elif f"end_{lbl}" in event and lbl in active:
                intervals.append((active.pop(lbl), ts, lbl))

    if not intervals:
        return []

    # 🔥 Convertir en temps relatif
    t0 = intervals[0][0]
    intervals = [(s - t0, e - t0, l) for s, e, l in intervals]

    print("  ℹ Intervalles (relatif):", intervals)
    return intervals


# ═══════════════════════════════════════════════════════════════════════════════
# 2. Resample → garder temps relatif
# ═══════════════════════════════════════════════════════════════════════════════
def resample_to_4hz(df, native_freq):
    df = df.copy()

    # 🔥 convertir en temps relatif
    df["timestamp"] = df["timestamp"] - df["timestamp"].iloc[0]

    df["datetime"] = pd.to_datetime(df["timestamp"], unit="ms")
    df = df.set_index("datetime").drop(columns=["timestamp"])

    period = "250ms"

    if native_freq > TARGET_FREQ:
        df = df.resample(period).mean()
    elif native_freq < 1:
        df = df.resample(period).ffill()
    elif native_freq < TARGET_FREQ:
        df = df.resample(period).interpolate()
    else:
        df = df.resample(period).mean()

    # 🔥 retour en ms relatif
    df["timestamp"] = (df.index - df.index[0]).total_seconds() * 1000

    return df.reset_index(drop=True)


# ═══════════════════════════════════════════════════════════════════════════════
# 3. Load + clean
# ═══════════════════════════════════════════════════════════════════════════════
def load_rename_resample(filepath, col_map, freq):
    try:
        df = pd.read_csv(filepath)
        df.columns = df.columns.str.strip().str.lower()

        ts_col = [c for c in df.columns if "time" in c][0]
        df = df.rename(columns={ts_col: "timestamp"})

        cols = {"timestamp": "timestamp"}
        for k, v in col_map.items():
            if k in df.columns:
                cols[k] = v

        df = df[list(cols.keys())].rename(columns=cols)
        df["timestamp"] = pd.to_numeric(df["timestamp"], errors="coerce")
        df = df.dropna().sort_values("timestamp")

        return resample_to_4hz(df, freq)

    except Exception as e:
        print("⚠ erreur:", filepath.name, e)
        return None


# ═══════════════════════════════════════════════════════════════════════════════
# 4. Breathing RPM
# ═══════════════════════════════════════════════════════════════════════════════
def compute_breathing_rpm(path):
    df = pd.read_csv(path)
    df.columns = df.columns.str.strip()

    ts = df["timestamp"].values
    sig = df["breathing_waveform"].values

    sig = sig - np.mean(sig)
    b, a = butter(2, [FMIN, FMAX], btype='band', fs=FS_BREATHING)
    sig = filtfilt(b, a, sig)

    peaks, props = find_peaks(
        sig,
        distance=int(MIN_DIST_S * FS_BREATHING),
        prominence=PROM_MIN
    )

    rows = []
    for i in range(len(peaks)-1):
        dt = (peaks[i+1] - peaks[i]) / FS_BREATHING
        rpm = 60 / dt

        rows.append({
            "timestamp": ts[peaks[i]],
            "breathing_rpm": rpm,
            "breathing_q": int(props["prominences"][i] >= PROM_OK)
        })

    df = pd.DataFrame(rows)

    return resample_to_4hz(df, 0.5)


# ═══════════════════════════════════════════════════════════════════════════════
# 5. Labels
# ═══════════════════════════════════════════════════════════════════════════════
def assign_labels(df, intervals):
    df["label"] = None

    for s, e, l in intervals:
        mask = (df["timestamp"] >= s) & (df["timestamp"] <= e)
        df.loc[mask, "label"] = l

    print("  ℹ labels trouvés:", df["label"].value_counts(dropna=False))

    return df.dropna(subset=["label"])


# ═══════════════════════════════════════════════════════════════════════════════
# MAIN
# ═══════════════════════════════════════════════════════════════════════════════
all_data = []

for p_dir in RAW_DIR.iterdir():
    if not p_dir.is_dir():
        continue

    for s_dir in p_dir.iterdir():
        if not s_dir.is_dir():
            continue

        print(f"\n▶ {p_dir.name} | {s_dir.name}")

        intervals = get_label_intervals(s_dir / "exp_markers.csv")
        if not intervals:
            continue

        merged = None

        # signals
        for f, cmap in FILE_COLS.items():
            path = s_dir / f
            if not path.exists():
                continue

            df = load_rename_resample(path, cmap, SENSOR_FREQ[f])
            if df is None:
                continue

            merged = df if merged is None else pd.merge_asof(
                merged.sort_values("timestamp"),
                df.sort_values("timestamp"),
                on="timestamp",
                tolerance=TARGET_PERIOD,
                direction="nearest"
            )
        # breathing
        b_path = s_dir / PATH_BREATHING
        if b_path.exists():
            df_b = compute_breathing_rpm(b_path)
            merged = pd.merge_asof(
                merged.sort_values("timestamp"),
                df_b.sort_values("timestamp"),
                on="timestamp",
                tolerance=TARGET_PERIOD,
                direction="nearest"
            )

        if merged is None:
            continue

        # DEBUG 🔥
        print("  🔍 timestamp range:", merged["timestamp"].min(), merged["timestamp"].max())

        merged = assign_labels(merged, intervals)

        if merged.empty:
            print("  ❌ toujours vide")
            continue

        merged["participant"] = p_dir.name
        merged["session"] = s_dir.name

        all_data.append(merged)

# ── SAVE ──────────────────────────────────────────────────────────────────────
if all_data:
    df = pd.concat(all_data)
    df.to_csv(OUTPUT_DIR / "dataset_clean.csv", index=False)

    print("\n✅ SUCCESS")
    print("Lignes:", len(df))
    print(df["label"].value_counts())
else:
    print("\n❌ Toujours aucune donnée (vérifier timestamps)")
````

````python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.preprocessing import RobustScaler

# ── Chemins ───────────────────────────────────────────────────────────────────
INPUT_PATH  = Path("/home/aziz/Desktop/fatigue_detection/edge-ai-wearable-fatigue-detection/data/02_interim/dataset_clean.csv")
OUTPUT_PATH = Path("/home/aziz/Desktop/fatigue_detection/edge-ai-wearable-fatigue-detection/data/02_interim/dataset_cleaned1.csv")

# ── Colonnes ──────────────────────────────────────────────────────────────────
COL_PARTICIPANT = "participant"
COL_SESSION     = "session"
COL_TIMESTAMP   = "timestamp"
COL_LABEL       = "label"

SIGNAL_COLS = [
    "acc_x", "acc_y", "acc_z",
    "eda", "wrist_hr", "ibi",
    "temp", "breathing_rpm"
]

LABEL_MAP = {
    "baseline": -1,
    "activity":  0,
    "fatigue":   1
}


# ═══════════════════════════════════════════════════════════════════════════════
# FONCTION 1 : Nettoyage
# ═══════════════════════════════════════════════════════════════════════════════
def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """
    Nettoie le dataset brut :
      - Supprime breathing_q (signal constant inutile)
      - Traite les NaN par ffill/bfill par groupe participant/session
      - Fallback médiane globale si NaN restants
      - Log transform sur EDA pour réduire l'asymétrie
      - Trie par participant puis session
    Retourne : df nettoyé (float64)
    """
    print("\n" + "=" * 60)
    print("ÉTAPE 1 : NETTOYAGE (clean_data)")
    print("=" * 60)

    df = df.copy()

    # 1.1 ── Supprimer breathing_q ────────────────────────────────────────
    if "breathing_q" in df.columns:
        df = df.drop(columns=["breathing_q"])
        print("✔ breathing_q supprimé")

    # 1.2 ── NaN par ffill/bfill par groupe ───────────────────────────────
    nan_before = df[SIGNAL_COLS].isna().sum().sum()
    print(f"\nNaN avant : {nan_before}")

    df[SIGNAL_COLS] = (
        df.groupby([COL_PARTICIPANT, COL_SESSION])[SIGNAL_COLS]
        .transform(lambda x: x.ffill().bfill())
    )

    nan_after = df[SIGNAL_COLS].isna().sum().sum()

    # Fallback médiane si NaN restants (groupe avec une seule valeur NaN)
    if nan_after > 0:
        print(f"⚠ {nan_after} NaN restants → médiane globale")
        for col in SIGNAL_COLS:
            if df[col].isna().sum() > 0:
                df[col] = df[col].fillna(df[col].median())

    print(f"✔ NaN traités : {nan_before} → {df[SIGNAL_COLS].isna().sum().sum()}")

    # 1.3 ── Log transform EDA ────────────────────────────────────────────
    skew_before = df["eda"].skew()
    df["eda"]   = np.log1p(df["eda"])
    print(f"✔ EDA log1p | asymétrie : {skew_before:.3f} → {df['eda'].skew():.3f}")

    # 1.4 ── Tri participant / session ────────────────────────────────────
    df = df.sort_values(
        by=[COL_PARTICIPANT, COL_SESSION],
        ascending=[True, True]
    ).reset_index(drop=True)
    print(f"✔ Trié par participant → session")

    print(f"\n✔ clean_data terminé | Shape : {df.shape}")
    return df


# ═══════════════════════════════════════════════════════════════════════════════
# FONCTION 2 : Encodage → séparation X / y
# ═══════════════════════════════════════════════════════════════════════════════
def encode_features(df: pd.DataFrame) -> tuple[pd.DataFrame, pd.Series]:
    """
    Encode le label et sépare features / target :
      - baseline → -1  |  activity → 0  |  fatigue → 1
      - Encodage dans la colonne 'label' (pas de colonne séparée)
      - Conversion float32 pour compatibilité STM32H7 (FPU 32-bit)
    Retourne : (X, y)
      X → DataFrame des signaux (float32)
      y → Series du label encodé (int8)
    """
    print("\n" + "=" * 60)
    print("ÉTAPE 2 : ENCODAGE (encode_features)")
    print("=" * 60)

    df = df.copy()

    # 2.1 ── Encoder le label ─────────────────────────────────────────────
    unknown = set(df[COL_LABEL].unique()) - set(LABEL_MAP.keys())
    if unknown:
        print(f"⚠ Labels inconnus : {unknown}")

    df[COL_LABEL] = df[COL_LABEL].map(LABEL_MAP)

    print(f"✔ Label encodé : baseline=-1 | activity=0 | fatigue=1")
    print(f"\n   Distribution :")
    print(df[COL_LABEL].value_counts().sort_index().to_string())

    # 2.2 ── Séparation X / y ─────────────────────────────────────────────
    X = df[SIGNAL_COLS].astype(np.float32)   # float32 → STM32H7 FPU natif
    y = df[COL_LABEL].astype(np.int8)         # int8    → économie mémoire

    print(f"\n✔ X shape : {X.shape} | dtype : {X.dtypes.unique()}")
    print(f"✔ y shape : {y.shape} | dtype : {y.dtype}")
    print(f"\n✔ encode_features terminé")
    return X, y


# ═══════════════════════════════════════════════════════════════════════════════
# FONCTION 3 : Normalisation
# ═══════════════════════════════════════════════════════════════════════════════
def normalize_features(
    X: pd.DataFrame,
    scaler: RobustScaler = None,
    fit: bool = True
) -> tuple[pd.DataFrame, RobustScaler]:
    """
    Normalise les features avec RobustScaler (robuste aux outliers, Q1/Q3) :
      - fit=True  → calcule le scaler sur X (à utiliser sur le train set)
      - fit=False → applique un scaler existant  (à utiliser sur le test set)
    Retourne : (X_normalized, scaler)
    ⚠ En LOSO : toujours fit=True sur le train, fit=False sur le test.
    """
    print("\n" + "=" * 60)
    print("ÉTAPE 3 : NORMALISATION (normalize_features)")
    print("=" * 60)

    X = X.copy()

    if fit:
        scaler = RobustScaler()
        X[SIGNAL_COLS] = scaler.fit_transform(X[SIGNAL_COLS])
        print(f"✔ RobustScaler fitted et appliqué")
    else:
        if scaler is None:
            raise ValueError("Fournir un scaler quand fit=False (ex: fold test en LOSO)")
        X[SIGNAL_COLS] = scaler.transform(X[SIGNAL_COLS])
        print(f"✔ RobustScaler appliqué (pré-calculé)")

    print(f"\n   Stats après normalisation :")
    print(X[SIGNAL_COLS].agg(["mean", "std"]).round(3).to_string())
    print(f"\n✔ normalize_features terminé | Shape : {X.shape}")
    return X, scaler


# ═══════════════════════════════════════════════════════════════════════════════
# PIPELINE PRINCIPAL
# ═══════════════════════════════════════════════════════════════════════════════
if __name__ == "__main__":

    print("=" * 60)
    print("CHARGEMENT DU DATASET")
    print("=" * 60)
    df = pd.read_csv(INPUT_PATH)
    print(f"✔ Dataset chargé : {df.shape}")

    # ── Pipeline en 3 étapes ──────────────────────────────────────────────
    df        = clean_data(df)
    X, y      = encode_features(df)
    X, scaler = normalize_features(X)

    # ── Sauvegarde ────────────────────────────────────────────────────────
    OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

    df_meta  = df[[COL_PARTICIPANT, COL_SESSION, COL_TIMESTAMP]].reset_index(drop=True)
    df_final = pd.concat([df_meta, X, y.rename(COL_LABEL)], axis=1)
    df_final.to_csv(OUTPUT_PATH, index=False)

    # ── Résumé ────────────────────────────────────────────────────────────
    print("\n" + "=" * 60)
    print("RÉSUMÉ FINAL")
    print("=" * 60)
    print(f"✔ Fichier sauvegardé : {OUTPUT_PATH}")
    print(f"✔ Shape finale       : {df_final.shape}")
    print(f"✔ NaN restants       : {df_final.isna().sum().sum()}")
    print(f"✔ RAM X (float32)    : {X.memory_usage(deep=True).sum() / 1024:.1f} KB")
    print(f"✔ Distribution label :")
    print(y.value_counts().sort_index().to_string())

    print("\n" + "=" * 60)
    print("⚠ RAPPEL LOSO — éviter le data leakage")
    print("=" * 60)
    print("""
  for test_subject in participants:
      df_train = df[df.participant != test_subject]
      df_test  = df[df.participant == test_subject]

      df_train         = clean_data(df_train)
      X_train, y_train = encode_features(df_train)
      X_train, scaler  = normalize_features(X_train, fit=True)

      df_test          = clean_data(df_test)
      X_test,  y_test  = encode_features(df_test)
      X_test,  _       = normalize_features(X_test, scaler=scaler, fit=False)
    """)
````
